```python
import numpy as np
import matplotlib.pyplot as plt
# this guy below uncommented makes plots pop out the notebook
%matplotlib tk
```

```python
class quadtree:
  def __init__(self, xmin, xmax, ymin, ymax, points, npoints_):
    self.bounds = (xmin, xmax, ymin, ymax)
    self.points = points
    self.children = []
    self.npoints_ = npoints_
    self.subdivide_if_needed()

  def subdivide_if_needed(self):
    if len(self.points) <= self.npoints_:
      return  # Leaf node

    xmin, xmax, ymin, ymax = self.bounds
    x_mid = (xmin + xmax) / 2
    y_mid = (ymin + ymax) / 2

    quadrants = [
      (xmin, x_mid, ymin, y_mid),  # Bottom-left
      (x_mid, xmax, ymin, y_mid),  # Bottom-right
      (xmin, x_mid, y_mid, ymax),  # Top-left
      (x_mid, xmax, y_mid, ymax),  # Top-right
    ]

    for (qx_min, qx_max, qy_min, qy_max) in quadrants:
      mask = (
        (self.points[:, 0] >= qx_min) & (self.points[:, 0] < qx_max) &
        (self.points[:, 1] >= qy_min) & (self.points[:, 1] < qy_max)
      )
      child_points = self.points[mask]
      # recursion base case
      if len(child_points) > 0:
        child_node = quadtree(qx_min, qx_max, qy_min, qy_max, child_points, self.npoints_)
        self.children.append(child_node)

  def get_leaf_nodes(self):
    if not self.children:
      return [self]
    leaves = []
    for child in self.children:
      leaves.extend(child.get_leaf_nodes())
    return leaves
# -----------------------------
def generate_points(n=1000, seed=42):
  np.random.seed(seed)
  return np.random.rand(n, 2)

def plot_clusters(leaves):
  colors = plt.cm.viridis(np.linspace(0, 1, len(leaves)))
  plt.figure(figsize=(8, 8))

  for i, node in enumerate(leaves):
    xmin, xmax, ymin, ymax = node.bounds
    rect = plt.Rectangle((xmin, ymin), xmax - xmin, ymax - ymin,
               edgecolor='black', facecolor='none', lw=1)
    plt.gca().add_patch(rect)
    points = node.points
    plt.scatter(points[:, 0], points[:, 1], s=10, color=colors[i], label=f'Cluster {i}')

  plt.title("Quadtree Clustering")
  plt.axis("equal")
  plt.show()
```

```python
points = generate_points(n=1000)
root = quadtree(0, 1, 0, 1, points, npoints_=5)
leaves = root.get_leaf_nodes()
print(f"Total clusters: {len(leaves)}")
plot_clusters(leaves)
```
